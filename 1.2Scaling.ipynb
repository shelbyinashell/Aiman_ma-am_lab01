```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/seishiro.jpeg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# image dimension
h, w = image.shape[:2] # (height, width), channel
```

```python
#scaling factors
sx = 5
sy = 3
```

```python
#scaling matrix
M = np.float32([
    [sx, 0, 0],
    [0, sy, 0]
])
```

```python
scaledImage= cv2.warpAffine(image, M, (int(w*sx),int(h*sy)))
```

```python
plt.figure(figsize=(10,5))
```

```python
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original image')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(scaledImage)
plt.title('Scaled image')
plt.axis('off')

plt.show()
```
