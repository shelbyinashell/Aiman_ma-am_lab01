```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/seishiro.jpeg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#Reflection about X axis
reflectedX = cv2.flip(image, 0)
#Reflection about Y axis
reflectedY = cv2.flip(image, 1)
```

```python
plt.figure(figsize=(40,37))
```

```python
plt.subplot(2,3,(1,3))
plt.imshow(image)
plt.title('Original image')
plt.axis('off')

plt.subplot(2,3,4)
plt.imshow(reflectedX)
plt.title('Reflection about X axis')
plt.axis('off')

plt.subplot(2,3,6)
plt.imshow(reflectedY)
plt.title('Reflection about Y axis')
plt.axis('off')

plt.show()
```
