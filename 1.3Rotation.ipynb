```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/seishiro.jpeg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
h, w = image.shape[:2]
```

```python
#Center of image
center= (w//2, h//2)
```

```python
#Rotation angle
angle= 50
```

```python
#Rotation matrix
M = cv2.getRotationMatrix2D(center, angle, 1.0)
```

```python
rotated= cv2.warpAffine(image, M, (w, h))
```

```python
plt.figure(figsize=(10,5))
```

```python
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original image')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(rotated)
plt.title('Rotated image - 50degree')
plt.axis('off')

plt.show()
```
